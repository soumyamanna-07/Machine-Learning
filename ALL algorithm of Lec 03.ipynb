```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset("titanic")  # In sns their is pre built data set that we can load that's why we are using
```

```python
df.head()
```

```python
# 1st we need to make ready the data
```

```python
df.columns
```

```python
df.info()
```

```python
# You can see their is too much null values in deck column , so we need to drop the columns
```

```python
df.drop([ 'embark_town', 'class', 'alive','who', 'adult_male', 'deck'],axis = 1,inplace = True)
```

```python
df.info()
```

```python
# now you can see in age , embarked there is some null or missing values
```

```python
df["age"].fillna(df["age"].mean(),inplace = True)
```

```python
# in embarked you can remove two null values's row instead of fill by mean or median or mode
```

```python
df.dropna(subset=["embarked"],inplace = True)
```

```python
df.info()
```

```python
# In here we predict survived or not
# so we can lebel encoding for catagorical values (whose has not int values)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df.head()
```

```python
df['sex'] = le.fit_transform(df['sex'])
df['embarked'] = le.fit_transform(df["embarked"])
```

```python

df = df.astype(int)
```

```python
df.head()
```

```python
# Here we can do scaling for age and fare to scale in one lebel, but here we will not apply , do apply direct logistic regression , we will do it in KNN
```

```python
# Let divide in X and Y lebel
x = df.drop("survived",axis = 1)     # as you know we will predict survived so we drop the columns form x

y = df["survived"]                     # replace in Y
```

```python
from sklearn.model_selection import train_test_split   # just split for train and test

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
# 2nd part is model evalution
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
# just import all things
```

```python
accuracy_score(y_test,y_pred)  # here is model accuracy
```

```python
confusion_matrix(y_test,y_pred)   # match with note book
```

```python
print(classification_report(y_test,y_pred))  # you can see all details about metrics evalution
```
